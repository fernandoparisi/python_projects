# **🚀NASA Turbofan Jet Engine**

### 🟢 **Información del Dataset**  - *Dataset Information*:

Este dataset proviene de la NASA y contiene datos simulados de degradación y falla (Run-to-Failure) de motores turbofan, generados con el simulador C-MAPSS bajo distintas condiciones operativas y modos de falla. El objetivo es predecir la **vida útil remanente (RUL)** de cada motor en el conjunto de prueba, es decir, cuántos ciclos de operación le quedan antes de fallar. Los datos son series temporales multivariadas de diferentes motores, con 3 configuraciones operativas y 26 variables (incluyendo sensores). El motor inicia en condiciones normales y desarrolla una falla progresiva. El dataset está dividido en subconjuntos de entrenamiento y prueba, con diferentes escenarios (FD001 a FD004) que varían en número de motores, condiciones y modos de falla.

**Fuente de datos:** https://www.kaggle.com/datasets/behrad3d/nasa-cmaps/data

**Subconjunto Utilizado:** Para nuestro análisis, hemos elegido el subconjunto **FD001** de los datos CMAPSS. Este subconjunto es ideal para un aprendizaje inicial debido a sus características simplificadas:

* **100 motores diferentes**: Esto proporciona una muestra manejable de comportamientos de motores.
* **1 condición operacional**: Todos los motores operan bajo las mismas condiciones de vuelo (altitud, velocidad, etc.).
* **1 modo de falla (HPC degradation)**: El único modo de falla presente en este subconjunto es la degradación del compresor de alta presión (HPC)

**Columnas:** Tanto el dataset de entrenamiento como el de prueba poseen las siguientes columnas:

* **`unit_number`**: Identificador único para cada motor (unidad). Cada `unit_number` representa un motor distinto que opera a lo largo de su vida útil.
* **`time_in_cycles`**: El tiempo en ciclos de operación para cada `unit_number`. Representa el progreso del motor a lo largo de su vida, donde un ciclo es una unidad de tiempo operativa.
* **`operational_setting_1`**: Primer configuración operativa del motor.
* **`operational_setting_2`**: Segunda configuración operativa del motor.
* **`operational_setting_3`**: Tercera configuración operativa del motor. 
* **`T2[°R]`**: Temperatura total en la entrada del fan (sensor 1). 
* **`T24[°R]`**: Temperatura total en la salida del fan (sensor 2).
* **`T30[°R]`**: Temperatura total en la salida del compresor de baja presión (LPC) (sensor 3).
* **`T50[°R]`**: Temperatura total en la salida de la turbina de alta presión (HPT) (sensor 4).
* **`P2[psia]`**: Presión en la entrada del fan (sensor 5). 
* **`P15[psia]`**: Presión total en la salida del fan (sensor 6).
* **`P30[psia]`**: Presión total en la salida del compresor de alta presión (HPC) (sensor 7).
* **`Nf[rpm]`**: Velocidad del rotor del fan (sensor 8).
* **`Nc[rpm]`**: Velocidad del rotor del compresor (sensor 9).
* **`epr[--]`**: Ratio de presión del bypass (sensor 10). 
* **`Ps30[psia]`**: Presión estática en la salida del compresor de alta presión (HPC) (sensor 11).
* **`phi[pps/psi]`**: Ratio de eficiencia del fan (sensor 12).
* **`NRf[rpm]`**: Velocidad corregida del rotor del fan (sensor 13).
* **`NRc[rpm]`**: Velocidad corregida del rotor del compresor (sensor 14).
* **`BPR[--]`**: Bypass Ratio (sensor 15).
* **`farB[--]`**: Ratio de combustible/aire en el quemador (sensor 16). 
* **`htBleed[--]`**: Purga de aire de alta presión (sensor 17).
* **`Nf_dmd[rpm]`**: Demanda de velocidad del rotor del fan (sensor 18). 
* **`PCNfR_dmd[rpm]`**: Demanda de velocidad corregida del rotor del fan (sensor 19). 
* **`W31[lbm/s]`**: Flujo de combustible (sensor 20).
* **`W32[lbm/s]`**: Flujo de combustible corregido (sensor 21).

*This dataset comes from NASA and contains simulated run-to-failure degradation data of turbofan engines, generated using the C-MAPSS simulator under various operating conditions and fault modes. The goal is to predict the **Remaining Useful Life (RUL)** of each engine in the test set, i.e., how many operating cycles remain before failure. The data consists of multivariate time series from different engines, with 3 operational settings and 26 variables (including sensors). Each engine starts under normal conditions and develops a progressive fault. The dataset is divided into training and test subsets, with different scenarios (FD001 to FD004) varying in number of engines, operating conditions, and fault modes.*

*Data source: https://www.kaggle.com/datasets/behrad3d/nasa-cmaps/data*

*Subset used:*
* _**100 different engines**_: _This provides a manageable sample of engine behaviors._
* _**1 operational condition**_: _All engines operate under the same flight conditions (altitude, speed, etc.)._
* _**1 fault mode (HPC degradation)**_: _The only fault mode present in this subset is High Pressure Compressor (HPC) degradation._

*Columns: Both the training and test datasets possess the following columns (prior to initial preprocessing steps like removing non-varying columns):*

* **`unit_number`**: _Unique identifier for each engine (unit). Each `unit_number` represents a distinct engine operating throughout its lifetime._

 #### ⚪ **Carga de todos los datasets**  - *All datasets upload*:

```python
# Definir los nombres de las columnas - Define the columns names
column_names = ['unit_number', 'time_in_cycles', 'setting_1', 'setting_2', 'setting_3', 
                'T2[°R]','T24[°R]','T30[°R]','T50[°R]','P2[psia]','P15[psia]','P30[psia]','Nf[rpm]','Nc[rpm]','epr[--]','Ps30[psia]','phi[pps/psi]','NRf[rpm]','NRc[rpm]','BPR[--]','farB[--]','htBleed[--]','Nf_dmd[rpm]','PCNfR_dmd[rpm]','W31[lbm/s]','W32[lbm/s]']

# Rutas de los archivos - File paths
train_data_path = 'https://raw.githubusercontent.com/fernandoparisi/python_projects/refs/heads/main/train_FD001.txt'
test_data_path = 'https://raw.githubusercontent.com/fernandoparisi/python_projects/refs/heads/main/test_FD001.txt'
rul_data_path = 'https://raw.githubusercontent.com/fernandoparisi/python_projects/refs/heads/main/RUL_FD001.txt'

# Columnas a eliminar (Sin variabilidad) - Columns to eliminate (Without variability)
# Basado en la matriz de correlación y EDA previo - Based on the correlation matrix and previous EDA
all_columns_to_drop = [
'setting_3',
    'T2[°R]',
    'P2[psia]',
    'epr[--]',
    'farB[--]',
    'Nf_dmd[rpm]',
    'PCNfR_dmd[rpm]'  
]

# Procesar df_train - Process df_train
df_train = pd.read_csv(train_data_path, sep=' ', header=None)
df_train.drop(columns=[26, 27], inplace=True) # Eliminar columnas vacías - Eliminate empty columns
df_train.columns = column_names

# Eliminar columnas irrelevantes de df_train - Remove irrelevant columns from df_train
df_train.drop(columns=all_columns_to_drop, inplace=True)
print(f"Columnas eliminadas de df_train: {all_columns_to_drop}")
print(f"Columnas finales de df_train: {df_train.columns.tolist()}")

# Calcular el RUL correcto para df_train - Calculate the correct RUL for df_train
max_cycles_train = df_train.groupby('unit_number')['time_in_cycles'].max().reset_index()
max_cycles_train.rename(columns={'time_in_cycles': 'max_cycle'}, inplace=True)
df_train = pd.merge(df_train, max_cycles_train, on='unit_number', how='left')
df_train['RUL'] = df_train['max_cycle'] - df_train['time_in_cycles'] # RUL real y decreciente - Real and decreasing RUL
df_train.drop(columns=['max_cycle'], inplace=True) # max_cycle ya no es necesario - max_cycle is not necessary

print("\nPrimeras filas de df_train con RUL calculado (debería ser decreciente):")
print(df_train[['unit_number', 'time_in_cycles', 'RUL']].head())
print("\nÚltimas filas del Motor 1 en df_train (RUL decreciente):")
print(df_train[df_train['unit_number'] == 1][['unit_number', 'time_in_cycles', 'RUL']].tail())
print(f"Dimensiones de df_train: {df_train.shape}\n")

# Procesar df_test - Process df_test
df_test = pd.read_csv(test_data_path, sep=' ', header=None)
df_test.drop(columns=[26, 27], inplace=True) # Eliminar columnas vacías - Eliminate empty columns
df_test.columns = column_names

# Eliminar las mismas columnas irrelevantes de df_test - Eliminate the same irrelevant columns from df_test
df_test.drop(columns=all_columns_to_drop, inplace=True)
print(f"Columnas eliminadas de df_test: {all_columns_to_drop}")
print(f"Columnas finales de df_test: {df_test.columns.tolist()}")

# Cargar y unificar RUL_true para df_test - Load and unify RUL_true for df_test
df_rul_true = pd.read_csv(rul_data_path, header=None, names=['RUL_true'])
df_rul_true['unit_number'] = range(1, len(df_rul_true) + 1) # Asignar unit_number - Assing unit_number 

# Calcular el RUL correcto para df_test (RUL decreciente en cada ciclo de test) - Calculate the correct RUL for df_test (RUL decreasing in each test cycle)
# 1. Obtener el último ciclo de cada motor en df_test - Get the last cycle for each motor in df_test
max_cycle_per_unit_test = df_test.groupby('unit_number')['time_in_cycles'].max().reset_index()
max_cycle_per_unit_test.rename(columns={'time_in_cycles': 'test_max_cycle'}, inplace=True)

# 2. Unir con df_test y luego con df_rul_true - Join with df_test y after with df_rul_true
df_test = pd.merge(df_test, max_cycle_per_unit_test, on='unit_number', how='left')
df_test = pd.merge(df_test, df_rul_true, on='unit_number', how='left')

# 3. Calcular la RUL para cada fila de df_test - Calculate the RUL for each row from df_test 
df_test['RUL'] = df_test['RUL_true'] + (df_test['test_max_cycle'] - df_test['time_in_cycles'])

# Limpiar columnas temporales - Clean temporal columns 
df_test.drop(columns=['RUL_true', 'test_max_cycle'], inplace=True)

print("\nPrimeras filas de df_test con RUL calculado (debería ser decreciente):")
print(df_test[['unit_number', 'time_in_cycles', 'RUL']].head())
print("\nÚltimas filas del Motor 1 en df_test (RUL decreciente):")
print(df_test[df_test['unit_number'] == 1][['unit_number', 'time_in_cycles', 'RUL']].tail())
print(f"Dimensiones de df_test: {df_test.shape}\n")
```

 #### ⚪ **Ingeniería de las características**  - *Feature Engineering*:

```python
# Identificar las columnas de sensores que quedaron después de la limpieza - Identify the columns of sensors which remaining after the cleaning:
fe_sensors = [col for col in df_train.columns if col.startswith('sensor_measurement_')]
windows_size = 20 # Puedes ajustar este tamaño de ventana

def apply_feature_engineering(df, sensors, window):
    df_fe = df.copy()
    for sensor in sensors:
        # Calcular media móvil - Calculate moving average
        df_fe[f'{sensor}_MA{window}'] = df_fe.groupby('unit_number')[sensor].rolling(
            window=window, min_periods=1).mean().reset_index(level=0, drop=True)
        
        # Calcular desviación estándar móvil - Calculate moving standard deviation
        df_fe[f'{sensor}_STD{window}'] = df_fe.groupby('unit_number')[sensor].rolling(
            window=window, min_periods=1).std().reset_index(level=0, drop=True)
        
        # Rellenar NaN que resultan de STD para las primeras observaciones - Fill NaN values resulting from STD for the first observations 
        df_fe[f'{sensor}_STD{window}'].fillna(0, inplace=True)
        
        # Calcular tasa de cambio usando la diferencia del ciclo anterior - Calculate rate of change using the difference from the previous cycle
        df_fe[f'{sensor}_DIFF'] = df_fe.groupby('unit_number')[sensor].diff().fillna(0) 
    return df_fe

df_train_fe = apply_feature_engineering(df_train, fe_sensors, windows_size)
df_test_fe = apply_feature_engineering(df_test, fe_sensors, windows_size)

print(f"\nColumnas de df_train_fe después de FE: {df_train_fe.columns.tolist()}")
print(f"Columnas de df_test_fe después de FE: {df_test_fe.columns.tolist()}")

print("\nPrimeras 5 filas de df_train_fe con nuevas features:")
print(df_train_fe.head())
print("\nPrimeras 5 filas de df_test_fe con nuevas features:")
print(df_test_fe.head())

# Preparar X e y para el modelo y escalado  - Prepare X and y for the model and scaling

# Definir las características a usar para el modelo - Define the features to use for the model
features_to_use = [col for col in df_train_fe.columns if col not in ['unit_number', 'time_in_cycles', 'RUL']]

print(f"\nCaracterísticas (features) seleccionadas para el modelo: {features_to_use}")
print(f"Cantidad de características: {len(features_to_use)}")

# Crear X e y - Create X and y
X_train = df_train_fe[features_to_use]
y_train = df_train_fe['RUL']

X_test = df_test_fe[features_to_use]
y_test = df_test_fe['RUL']

print(f"\nDimensiones de X_train: {X_train.shape}, y_train: {y_train.shape}")
print(f"Dimensiones de X_test: {X_test.shape}, y_test: {y_test.shape}")

# Escalado de características - Feature scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convertir de vuelta a DataFrame para mantener los nombres de las columnas - Convert back to DataFrame to preserve column names
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)

print(f"\nPrimeras filas de X_train_scaled_df después de escalado:\n{X_train_scaled_df.head()}")
print(f"\nPrimeras filas de X_test_scaled_df después de escalado:\n{X_test_scaled_df.head()}")
```

```python
import xgboost as xgb 
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Inicializar y entrenar el modelo XGBoost Regressor - Initialize and train the XGBoost Regressor model
# Hiperparámetros comunes para empezar - Common hyperparameters to get started :
# n_estimators: número de árboles - number of trees
# learning_rate: tamaño del paso en cada iteración - step size in each iteration
# max_depth: profundidad máxima de los árboles - maximum depth of trees
# n_jobs: para paralelización (-1 usa todos los núcleos disponibles) - for parallelization (-1 uses all available cores)
# random_state: para reproducibilidad - for reproducibility
# objective: 'reg:squarederror' para regresión - 'reg:squarederror' for regression
xgb_model = xgb.XGBRegressor(n_estimators=100, learning_rate=0.1, max_depth=3,
                             n_jobs=-1, random_state=42, objective='reg:squarederror')
xgb_model.fit(X_train_scaled, y_train)

# Realizar predicciones en el conjunto de entrenamiento - Make predictions on the training dataset
y_train_pred_xgb = xgb_model.predict(X_train_scaled)

# Realizar predicciones en el conjunto de prueba - Make predictions on the test dataset
y_test_pred_xgb = xgb_model.predict(X_test_scaled)

# Asegurarse de que las predicciones de RUL no sean negativas - Ensure RUL predictions are not negative
y_train_pred_xgb[y_train_pred_xgb < 0] = 0
y_test_pred_xgb[y_test_pred_xgb < 0] = 0

print("\n--- Evaluación del Modelo XGBoost en el Conjunto de Entrenamiento - Evaluating the XGBoost Model on the Training Set ---")
mae_train_xgb = mean_absolute_error(y_train, y_train_pred_xgb)
mse_train_xgb = mean_squared_error(y_train, y_train_pred_xgb)
rmse_train_xgb = np.sqrt(mse_train_xgb)
r2_train_xgb = r2_score(y_train, y_train_pred_xgb)

print(f"MAE (Train XGB): {mae_train_xgb:.4f}")
print(f"MSE (Train XGB): {mse_train_xgb:.4f}")
print(f"RMSE (Train XGB): {rmse_train_xgb:.4f}")
print(f"R2 Score (Train XGB): {r2_train_xgb:.4f}")

print("\n--- Evaluación del Modelo XGBoost en el Conjunto de Prueba - Evaluating the XGBoost Model on the Test Set ---")
mae_test_xgb = mean_absolute_error(y_test, y_test_pred_xgb)
mse_test_xgb = mean_squared_error(y_test, y_test_pred_xgb)
rmse_test_xgb = np.sqrt(mse_test_xgb)
r2_test_xgb = r2_score(y_test, y_test_pred_xgb)

print(f"MAE (Test XGB): {mae_test_xgb:.4f}")
print(f"MSE (Test XGB): {mse_test_xgb:.4f}")
print(f"RMSE (Test XGB): {rmse_test_xgb:.4f}")
print(f"R2 Score (Test XGB): {r2_test_xgb:.4f}")

# Visualización de Resultados para XGBoost - Visualizing Results for XGBoost
num_units_to_plot_xgb = 6
test_unit_numbers_xgb = df_test['unit_number'].unique()[:num_units_to_plot_xgb]
nrows = 2
ncols = 3

plt.figure(figsize=(25, 10)) 

for i, unit_num in enumerate(test_unit_numbers_xgb):
    ax = plt.subplot(nrows, ncols, i + 1)

    unit_mask = (df_test['unit_number'] == unit_num)
    actual_rul_xgb = y_test[unit_mask]
    predicted_rul_xgb = y_test_pred_xgb[unit_mask] 
    time_cycles_xgb = df_test[unit_mask]['time_in_cycles']

    ax.plot(time_cycles_xgb, actual_rul_xgb, label='RUL Real - Real RUL', color='blue')
    ax.plot(time_cycles_xgb, predicted_rul_xgb, label='RUL Predicho (XGB) - Predicted RUL (XGB)', color='gray', linestyle='--') 
    ax.set_title(f'Unidad {unit_num} - Unit {unit_num}')
    ax.set_xlabel('Ciclos - Cycles')
    ax.set_ylabel('RUL')
    ax.legend(fontsize=8)
    ax.grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.90])
plt.suptitle('RUL Real vs. Predicho por XGBoost para Motores de Prueba Seleccionados\n'
             'Real RUL vs. Predicted by XCGBoost for Selected Test Engines', fontsize=16)
plt.show()

# Gráfico de Dispersión para XGBoost - Scatter Plot for XGBoost
plt.figure(figsize=(8, 4)) 
plt.scatter(y_test, y_test_pred_xgb, alpha=0.3, color='gray') 
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label='Predicción Perfecta - Perfect Prediction')
plt.xlabel('RUL Real - Real RUL')
plt.ylabel('RUL Predicho (XGB) - Predicted RUL (XGB)')
plt.title('RUL Real vs. Predicho (XGBoost - Conjunto de Prueba)\n'
          'Real RUL vs. Predicted (XGBoost - Test Dataset)')
plt.grid(True)
plt.legend()
plt.show()
```
